```python
from pathlib import Path

# CHANGE THIS
MUSIC_FOLDER = Path("/Users/hassi/Projects/Psychoacoustic-Parameter-Estimation-via-ML/data/music_dataset_jamendo_split/60 min")

RECURSIVE = True
```

change suffix: .html -> .mp3

```python
html_files = (
    list(MUSIC_FOLDER.rglob("*.html"))
    if RECURSIVE
    else list(MUSIC_FOLDER.glob("*.html"))
)

for old_path in html_files:
    new_path = old_path.with_suffix(".mp3")
    old_path.rename(new_path)
    print(f"{old_path.name} -> {new_path.name}")

print(f"\nRenamed {len(html_files)} files.")
```

calculate duration and disc space

```python
from mutagen import File as MutagenFile

audio_extensions = {
    ".mp3", ".wav", ".flac", ".m4a",
    ".aac", ".ogg", ".opus", ".aiff", ".aif"
}

audio_files = [
    p for p in MUSIC_FOLDER.rglob("*")
    if p.is_file() and p.suffix.lower() in audio_extensions
]

total_seconds = 0
total_bytes = 0

for file in audio_files:
    total_bytes += file.stat().st_size

    try:
        audio = MutagenFile(file)
        if audio and audio.info:
            total_seconds += audio.info.length
    except Exception:
        pass

print(f"Files: {len(audio_files)}")
print(f"Duration: {total_seconds/60:.2f} min ({total_seconds/3600:.2f} h)")
print(f"Data: {total_bytes/1024**3:.2f} GiB")
print(f"Data: {total_bytes/1e9:.2f} GB")
```
